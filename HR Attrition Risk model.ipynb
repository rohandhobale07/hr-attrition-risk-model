```python
x_cat = df[['BusinessTravel', 'Department', 'EducationField', 'Gender', 'JobRole', 'MaritalStatus']]
x_cat
```

```python
from sklearn.preprocessing import OneHotEncoder
onehotencoder = OneHotEncoder()
x_cat = onehotencoder.fit_transform(x_cat).toarray()
x_cat
```

```python
x_cat=pd.DataFrame(x_cat)
x_cat
```

```python
x_numerical = df[['Age', 'DailyRate', 'DistanceFromHome',	'Education', 'EnvironmentSatisfaction', 'HourlyRate', 'JobInvolvement',	'JobLevel',	'JobSatisfaction',	'MonthlyIncome',	'MonthlyRate',	'NumCompaniesWorked',	'OverTime',	'PercentSalaryHike', 'PerformanceRating',	'RelationshipSatisfaction',	'StockOptionLevel',	'TotalWorkingYears'	,'TrainingTimesLastYear'	, 'WorkLifeBalance',	'YearsAtCompany'	,'YearsInCurrentRole', 'YearsSinceLastPromotion',	'YearsWithCurrManager']]
display(x_numerical)
```

```python
x_all=pd.concat([x_cat, x_numerical], axis=1)
x_all
```

```python
x_all.columns=x_all.columns.astype(str)
```

```python
x = x_all.values
x
```

```python
y = df['Attrition']
y.value_counts()
```

```python
from sklearn.preprocessing import MinMaxScaler
mms = MinMaxScaler()
x = mms.fit_transform(x_all)
x

```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x, y, test_size=0.30, random_state=7)
print(x.shape)
print(y.shape)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_train, y_train)
```

```python
y_pred=lr.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_test, y_pred)
acc
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test, y_pred)
cm
```

```python
from sklearn.metrics import classification_report
cr=classification_report(y_test, y_pred)
print(cr)
```

```python
from sklearn.metrics import recall_score

lr_bal = LogisticRegression(class_weight='balanced', max_iter=1000)
lr_bal.fit(x_train, y_train)
y_pred_bal = lr_bal.predict(x_test)

print('--- Balanced Logistic Regression ---')
print('Accuracy:', accuracy_score(y_test, y_pred_bal))
print(classification_report(y_test, y_pred_bal))
```

```python
from sklearn.ensemble import RandomForestClassifier
rfc=RandomForestClassifier()
rfc.fit(x_train, y_train)
```

```python
y_pred=rfc.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_test, y_pred)
acc
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test, y_pred)
cm
```

```python
from sklearn.metrics import classification_report
cr=classification_report(y_test, y_pred)
print(cr)
```

```python
rfc_bal = RandomForestClassifier(class_weight='balanced', random_state=42)
rfc_bal.fit(x_train, y_train)
y_pred_rfc_bal = rfc_bal.predict(x_test)

print('--- Balanced Random Forest ---')
print('Accuracy:', accuracy_score(y_test, y_pred_rfc_bal))
print(classification_report(y_test, y_pred_rfc_bal))
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc=DecisionTreeClassifier()
dtc.fit(x_train, y_train)
```

```python
y_pred=dtc.predict(x_test)
y_pred
```

```python
from sklearn.metrics import accuracy_score
acc=accuracy_score(y_test, y_pred)
acc
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test, y_pred)
cm
```

```python
from sklearn.metrics import classification_report
cr=classification_report(y_test, y_pred)
print(cr)
```

```python
summary = pd.DataFrame({
    'Model': ['Logistic Regression (baseline)', 'Logistic Regression (balanced)', 'Random Forest (balanced)'],
    'Accuracy': [
        accuracy_score(y_test, lr.predict(x_test)),
        accuracy_score(y_test, y_pred_bal),
        accuracy_score(y_test, y_pred_rfc_bal)
    ],
    'Recall (Attrition=Yes)': [
        recall_score(y_test, lr.predict(x_test), pos_label=1),
        recall_score(y_test, y_pred_bal, pos_label=1),
        recall_score(y_test, y_pred_rfc_bal, pos_label=1)
    ]
})
display(summary)

print('\nKey finding: Balanced Logistic Regression trades ~12 points of accuracy for a')
print('near-doubling of recall on the attrition class (40% -> 70%) - the more useful')
print('trade-off for an HR use case where missing at-risk employees is costly.')
```
